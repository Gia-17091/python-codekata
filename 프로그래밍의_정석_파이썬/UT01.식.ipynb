```python
# 연습문제 1.1] 초고속 열차 시속 계산
# 서울~부산 325km 16분 개발 -> 현실이 되려면, 시속 km?

answer = 325 * 60 / 16
answer = int(answer)
print(f"{answer} km/h")
```

```python
# 연습문제 1.2] 마라톤 세계 기록의 초속 계산
# 마라톤: 42.195 km
# 2018년: 2시간 1분 39초
# 평균: 1초당 몇 미터?

m = 2 * 60 + 1
s = m * 60 +39

km = 42.195
mi = km * 1000

answer = mi / s
print(f"평균 {int(answer)}미터")
```
