````python
# 기말 평균 학점 구하기
# 과목별 이수 학점
python = 3
mobile = 2
excel = 1

# 성적
A = 4.5
A0 = 4.0
B = 3.5

# 이수 학점의 합
subjects = 0
subjects += python
subjects += mobile
subjects += excel

# 과목별 성적
sum_sub = 0
sum_sub += python * B
sum_sub += mobile * A0
sum_sub += excel * A

avg_sub = sum_sub / subjects
print("평균 학점: ", avg_sub)
````

### **조건문**
- if-else문
if 조건식:
    실행할 문장
=> 자동으로 `4칸`이 뛰어짐.

- 들여쓰기로, 중첩 if문 등 구분.

````python
# [암기] 짝수/홀수 판별 - 두 수 입력 (첫 짝수: 합 /홀수: 빼기)
num1 = int(input("숫자를 입력하시오: "))
num2 = int(input("숫자를 입력하시오: "))

if num1 % 2 == 0:
    print(num1, "은 짝수 입니다.")
    print("입력하신 두 수의 합을 구하겠습니다.")
    answer = num1 + num2
    print("두 수의 합은 ", answer, "입니다.")
else:
    print(num1, "은 홀수 입니다.")
    print("입력하신 두 수의 차를 구하겠습니다.")
    answer = num1 - num2
    print("두 수의 차는 ", answer, "입니다.")
````

````python
# 성적을 입력해 학점 구하기
score = int(input("점수를 입력하시오: "))

if score >= 90:
    print("A", end='')
elif score >= 80:
    print("B", end='')
elif score >= 70:
    print("C", end='')
elif score >= 60:
    print("D", end='')
else:
    print("F", end='')
print("학점입니다.")
````

````python
# 사용자로부터 정수를 입력받아 3의 배수 혹은 5의 배수이지 판별
int1 = int(input("정수를 입력하시오: "))

if int1 % 3 ==0:
    if int1 % 5 == 0:
        print("Python-DS")
    else:
        print("Python")
elif int1 % 5 == 0:
    print("DS")
else:
    print(int1, "은 해당 없음.")
    
print("수업 끝!")
````
