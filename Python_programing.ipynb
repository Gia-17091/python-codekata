### 1. 토큰과 주석

- 한국어로 작성된 글이 단어로 이루어져 있듯이, 파이썬 프로그램은 **토큰**(token)이라고 하는 기본 요소들로 구성
- 파이썬 언어에서 사용되는 토큰은 키워드, 식별자, 리터럴, 연산자, 구분자의 다섯 가지로 분류
- **키워드**(keywords)는 파이썬 프로그램에서 정해진 역할을 수행하도록 *약속되어 있는* 35개의 단어
  - `True` `False` `None`
  - **`and`** `as` `assert` `async` `await` `break` `class` `continue` `def` `del` `elif`
  - `else` `except` `finally` `for` `from` `global` `if` `import` `in` **`is`** `lambda`
  - `nonlocal` **`not`** **`or`** `pass` `raise` `return` `try` `while` `with` `yield`
- **식별자**(identifiers)는 프로그래머가 어떤 값이나 대상에 부여하는 이름
  - *문자*와 `_` 또는 *숫자*의 조합으로 이루어져 있으면서 키워드가 아닌 것
  - 단, 식별자는 숫자로 시작할 수 없으므로 첫 글자는 반드시 영문, 한글 등의 문자 또는 `_`이어야 함
  - `n` `String` `DATA` `x1` `quick_sort` `__name__` `_` 등은 모두 식별자를 나타냄
  - 규칙을 만족하더라도 문자 리터럴이나 주석에 포함되어 있는 부분은 식별자가 아님
- **리터럴**(literals)은 고정된 값이나 대상을 나타내기 위해 사용하는 표현
  - 대문자로 시작하는 키워드인 `True` `False` `None`은 특수 상수(singleton) 리터럴로 분류
  - 숫자로 시작하는 표현은 숫자 리터럴이라고 하며 수를 나타내기 위해 사용: `1` `3.14` `1.6e-19` `3.4j` 등
  - 따옴표로 둘러쌓인 표현은 문자 리터럴이라고 하며 텍스트를 나타내기 위해 사용: `'Hello'` `"Hello, World!"` 등
- **연산자**(operators)는 규칙에 따라 값을 계산하도록 약속되어 있는 기호
  - 특수 문자 `+` `-` `*` `**` `/` `//` `%` `@` `&` `|` `^` `<<` `>>` `~` `<` `>` `<=` `>=` `==` `!=` `:=` 은 연산자
  - 일부 키워드(`and` `or` `not` `is`) 및 키워드의 조합(`is not`)도 연산자
  - 또 다른 일부 키워드 및 키워드의 조합(`in` `not in` `if else`)은 때때로 연산자
- **구분자**(delimiters)는 그 외에 특별한 역할을 하는 기호
  - `=` `+=` `-=` `*=` `**=` `/=` `//=` `%=` `@=` `&=` `|=` `^=` `<<=` `>>=` `( )` `[ ]` `{ }` `;` `:` `,` `.` `->` `@`
  - `=` `+=` 등은 파이썬에서 연산자가 아니지만 다른 언어에서 연산자이므로 관례에 따라 연산자라고 부르기도 함
  - `( )`는 일반적으로 구분자로 분류되지만, 객체를 호출하는 용도로 사용될 때에는 *호출 연산자*라고 부름

##### [Prob] 다음 프로그램에서 키워드, 식별자, 리터럴, 연산자, 구분자, 그리고 주석을 찾아 보자.

```python
# Printing heart 프로그램
y = 1.3
while y > -1.05:
    x = -1.1
    while x < 1.15:
        if (x**2 + y**2 - 1)**3 <= x**2 * y**3: # 좌표가 heart 안쪽에 해당하는 경우
            print('*', end = ' ')
        else:
            print(' ', end = ' ')
        x += 0.1
    y -= 0.1
    print()
```

### 2. 객체, 타입, 클래스, 인스턴스

- 파이썬에서는 프로그램이 실행되는 동안 메모리에 저장되는 모든 대상을 **객체**(object)라고 부름
- 프로그램에서 객체를 참조하기 위해서는 객체를 가리키는 표현(토큰)이 필요한데, 그것이 바로 리터럴과 식별자
- 리터럴은 가리키는 대상이 고정되어 있으므로 항상 타입이 일정하지만, 식별자는 가리키는 대상에 따라 타입이 달라짐
- 파이썬 프로그램에서는 연산자와 구분자, 그리고 소문자로 시작하는 32개의 키워드를 제외한 토큰은 모두 객체
  - `x = 1 + 2`라는 표현에 등장하는 `x`, `1`, `2`는 모두 객체를 지칭하는 표현
- 객체는 **자료형**(data type)에 따라 분류가 가능한데, 자료형을 다른 말로 **타입**(type) 또는 **클래스**(class)라고 함

- 어떤 타입의 객체는 해당 **클래스**(class)의 **인스턴스**(instance)라고 표현함

### 3. 수치 자료형과 숫자 리터럴

- 고정된 값을 갖는 수를 나타내기 위한 토큰을 **숫자 리터럴**이라고 함
- 숫자 리터럴이 표현하는 객체는 모두 `int`, `float` 또는 `complex` 타입에 속함
- `0`부터 `9`까지의 숫자만으로 표현되는 **정수 리터럴**은 `int` 타입 객체를 나타내며, 정수를 표현하기 위해 사용
    - `0`, `1`, `12`, `2147483648`, `18446744073709551616` 등
    - 다른 언어와 달리 파이썬에서는 표현 가능한 정수의 범위가 정해져 있지 않아서 매우 큰 정수도 사용 가능함
- 파이썬 3.6 버전부터 숫자 리터럴의 숫자와 숫자 사이에 자릿수 구분을 도와주는 `_`을 추가할 수 있도록 바뀜
    - 파이썬 3.6 이상 버전에서는 `1234567890`, `1_234_567_890`, `12_3456_7890`가 모두 같은 값을 나타냄

```python
type(12_3456_7890)
```

- 소수점 `.` 또는 `E` 또는 `e`가 포함된 **실수 리터럴**은 `float` 타입의 객체를 나타내며, 실수를 표현하기 위해 사용
    - `float`은 floating point number의 약자로 $12 = 1.2 \times 10 = 0.12 \times 10^2$처럼 소수점 위치가 변하는 수라는 의미
    - `0.0`, `1.0`, `3.14`, `12e34`, `1.6E-19`
    - 참고로 `e`는 지수를 뜻하는 exponent의 첫 글자로, `12e34`는 $12 \times 10^{34}$을 의미함
    - 소수점 앞 또는 뒤에 `0` 하나만 있는 경우, `0` 하나를 생략 가능
    - 예를 들어 `0.0`은 `.0` 또는 `0.`으로 표기 가능하며, `1.0`은 `1.`으로, `0.1`은 `.1`로 표기 가능
    - 수학에서 `1`과 `1.0`은 동일한 수를 나타내지만 파이썬에서는 타입이 다르므로 서로 다른 객체를 나타냄

```python
type(1)
```

```python
type(1.0)
```

- 숫자로 시작해서 `J` 또는 `j`로 끝나는 표현은 **허수 리터럴**로, 복소수의 허수부를 표현하기 위해 사용
    - `j`와 `J`는 단위 허수 $i = \sqrt{-1}$ 을 나타냄
    - 허수 리터럴로 표현된 객체는 `complex` 타입으로, 실수 및 허수 리터럴의 합과 차를 사용해 임의의 복소수 표현
    - `1j`, `1`+`2J`, `1.0`-`2.0j`, `1.2e18`+`1.6e-19J`
    - 예를 들어 `1.2e18`+`1.6e-19J`는 복소수 $(1.2\times10^{18}) + i (1.6\times10^{-19})$을 의미

```python
type(1.2e18 + 1.6e-19j)
```

- `'Hello'`와 `"Hello"`는 둘 다 다섯 개의 문자 `H`, `e`, `l`, `l`, `o`를 모아 놓은 동일한 문자열을 의미

### 5. 식과 문장

- 파이썬 언어의 토큰들을 적절히 조합하면 **식**(expression)과 **문장**(statement)이 만들어짐
- 파이썬에서 **연산자**와 **피연산자**를 규칙에 따라 조합하면 식이 만들어지며, 모든 식은 연산의 결과로 어떤 값을 만들어냄
- 연산자의 종류에 따라 1~3개의 피연산자가 조합되는데, 피연산자의 자리에는 식별자나 리터럴 또는 다른 식이 올 수 있음
- 예를 들어 연산자 `+`에 리터럴 `1`과 `2`를 조합하여 만들어지는 `1 + 2`라는 표현은 `3`이라는 결과 값을 만드는 식
- 식을 계산하여 값을 얻는 과정을 식을 **평가**(evaluate)한다고 표현함
- 파이썬에서는 **평가**(evaluation)하면 어떤 결과 값(객체)을 도출하도록 약속되어 있는 모든 표현을 **식**(expression)이라고 함
  - 하나의 식별자
  - 하나의 리터럴
  - 연산자와 피연산자를 규칙에 따라 결합한 표현
  - 함수를 호출하는 표현

##### [Prob] 다음 프로그램에서 식을 찾아 보자.

```python
# Printing heart 프로그램
y = 1.3
while y >= -1.1:
    x = -1.1
    while x < 1.1:
        if (x**2 + y**2 - 1)**3 <= x**2 * y**3:
            print('*', end=' ')
        else:
            print(' ', end=' ')
        x += 0.1
    y -= 0.1
    print()
```

- 실행 가능한 최소한의 코드 조각을 **문장**이라고 하며, 흔히 코드상의 하나의 줄로 표현
- 예를 들어 `y = 1.3`은 식별자 `y`로 표현되는 변수에 리터럴 `1.3`으로 표현되는 객체를 대입하도록 지시하는 문장
- 식은 모두 그 자체로 실행 가능하므로 모두 문장이지만, 문장 중에는 식이 아닌 것도 있음
  - `1 + 2`는 1과 2를 더해서 그 결과 값을 얻도록 지시하는 문장이고, `3`이라는 결과 값을 가지므로 동시에 식
  - `x = 1`은 그 자체로 실행 가능하므로 문장이지만 결과가 나타내는 값이 없으므로 식이 아님
  - 그러나 해당 문장 안에 들어 있는 `x` 또는 `1`은 메모리에 저장된 어떤 값(객체)을 나타내므로 식
- 어떤 문장은 다른 문장들을 포함하기도 하는데, 이러한 문장을 **복합문**이라고 하며 보통 코드상의 여러 줄로 표현
  - 예를 들어 `if` 키워드로 시작하는 조건문, `while` 또는 `for` 키워드로 시작하는 반복문은 모두 복합문
  - 또한 `def` 키워드로 시작하는 함수 정의문, `class` 키워드로 시작하는 클래스 정의문 등도 모두 복합문
  - 복합문에 포함되는 문장들은 들여쓰기로 구분하며, 동일한 만큼 들여쓰기 된 문장들은 모두 하나의 복합문에 포함

### 6. 수식과 산술 연산자

- **수식**은 수치 연산을 위한 **산술 연산자**를 피연산자와 정해진 규칙에 따라 결합하여 만들어지는 식
- 파이썬에는 총 9가지의 산술 연산자가 있음
  - `+`, `-`, `*`, `/`은 수학의 사칙 연산(덧셈, 뺄셈, 곱셈, 나눗셈)을 의미
  - `//`과 `%`는 나눗셈의 몫과 나머지를 의미
  - `**`은 거듭제곱을 의미
- 대부분의 산술 연산자는 왼쪽에 하나 오른쪽에 하나 이렇게 두 개의 피연산자를 필요로 하는 **이항 연산자**
- `+`와 `-`는 오른쪽에만 피연산자가 있는 **단항 연산자**로 사용되기도 함
  - 단항 `-` 연산자는 오른쪽 피연산자의 부호를 반대로 바꾼 값을 줌
  - 단항 `+` 아무 것도 하지 않고 오른쪽 피연산자의 값을 그대로 줌

| 연산       | 연산 식의 결과 값       | 우선 순위 | 결합 순서 |

| `x // y` | $x$를 $y$로 나눈 몫     |  2        | 좌결합    |

| `x - y`  | $x$에서 $y$를 뺀 값     |  3        | 좌결합    |

##### [Prob] 임시 변수 `t`를 사용해 변수 `x`와 `y`에 대입되어 있는 값을 서로 교환하는 문장을 작성해 보자.

```python


print('x = ', x, ', y = ', y, sep='')
```

- 파이썬처럼 변수의 타입이 저장하는 대상에 따라 달라지는 언어를 **동적 타입**(dynamic typing) 언어라고 함
- 하나의 노트북 파일에서 안에서는 한 셀에서 변수에 값을 대입하면 그 변수를 다른 셀에서 계속 이어서 사용 가능
- 파이썬에서는 아래와 같은 표현을 사용하여 여러 변수에 같은 값을 일괄적으로 대입하는 것이 가능함

```python
x = y = z = 1
print(x, y, z)
```

- 또한, 여러 변수에 서로 다른 값을 한 번에 대입할 때에는 아래와 같은 표현을 사용 가능함

```python
x, y, z = 1, 2, 3
print(x, y, z)
```

- 이를 활용하면 임시 변수를 사용하지 않고도 변수 `x`와 `y`에 대입되어 있는 값을 교환할 수 있음

```python
x, y = y, x
print('x = ', x, ', y = ', y, sep='')
```

- 파이썬에서 `+=` `-=` `*=` `**=` `/=` `//=` `%=` 등의 구분자는 변수에 저장된 값을 변화시키는 일을 함
- 예를 들어서 대입문 `x = x + 1`은 위 구분자를 사용하여 `x += 1`로 좀더 함축적으로 표시할 수 있음

##### [Prob] 변수 `z`에 대입된 값을 1만큼 감소시키는 문장을 작성해 보자.

- 파이썬에서 `x = 1`과 같은 대입문은 표현 자체가 어떤 값을 나타내는 것이 아니므로 식이 아님

```python
y = (x = 1) + 2
```

- 반면에 다른 많은 언어에서 `x`에 `1`을 대입하는 동작을 수행하는 동시에 대입된 값 자체를 나타내는 식
- 그래서 C 언어 등에서는 `=`을 대입 연산자 `+=` `-=` 등은 복합 대입 연산자라고 부름
- 파이썬에도 3.8 버전부터 다른 언어의 대입 연산자 `=`와 유사한 기능을 하는 **왈러스 연산자** `:=`가 도입됨
- 아래 코드의 `x := 1`은 변수 `x`에 `1`을 대입함과 동시에 그 자체로 `x`에 대입된 객체를 나타내는 식

```python
y = (x := 1) + 2
print(x)
print(y)
```

- 대입문 대신 왈러스 연산자를 사용한 **대입식**을 사용하면 코드를 좀더 간결하게 작성할 수 있는 경우가 있음
- 구체적인 예는 반복문을 사용한 입력 유효성 검사 관련 내용을 다루면서 설명할 예정
